```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Configuration for visual style
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 10})


# 1. Customer Status Distribution (Figure 2.1)
df_status = pd.DataFrame({
    'Status': ['ACTIVE', 'SUSPENDED', 'PENDING', 'DORMANT', 'CLOSED'],
    'Count': [98420, 18340, 12760, 8410, 4450]
})

# 2. Customer Type Distribution (Figure 2.2)
df_type = pd.DataFrame({
    'Type': ['INDIVIDUAL', 'CORPORATE', 'SME', 'NONPROFIT', 'GOVERNMENT'],
    'Count': [88500, 27400, 16900, 6400, 3180]
})

# 3. Gender Distribution (Figure 3.1)
df_gender = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Non-Binary', 'Prefer Not to Say'],
    'Count': [67840, 58920, 9240, 6380]
})

# 4. Age Group Distribution (Figure 3.2)
df_age = pd.DataFrame({
    'Age_Group': ['Under 18', '18-25', '26-35', '36-45', '46-60', '60+'],
    'Count': [2840, 24600, 42100, 31800, 27400, 13640]
})

# 5. Top 10 Countries by Count (Figure 4.1)
df_geo = pd.DataFrame({
    'Country': ['USA', 'UK', 'India', 'Germany', 'Canada', 'Australia', 'France', 'Brazil', 'Japan', 'South Africa'],
    'Count': [38200, 19400, 17600, 11300, 9800, 7400, 6900, 6100, 5200, 4100]
})

# 6. Income Band Distribution (Figure 5.2)
df_income = pd.DataFrame({
    'Band': ['< $25k', '$25k–$50k', '$50k–$100k', '$100k–$200k', '$200k+'],
    'Count': [18400, 31200, 42600, 21800, 10600]
})

# 7. Proof of Identity Types (Figure 6.2)
df_id = pd.DataFrame({
    'Doc_Type': ['Passport', 'Driver License', 'Utility Bill', 'Bank Statement', 'Gov. Letter'],
    'Count': [41200, 34600, 22400, 18900, 9280]
})

# 8. Profile Completeness Scores (Figure 9.1)
df_complete = pd.DataFrame({
    'Score': ['1 (Very Low)', '2', '3', '4', '5', '6 (Complete)'],
    'Count': [4200, 8400, 18600, 32400, 41800, 36980]
})

# --- PLOTTING ALL CHARTS ---
fig, axes = plt.subplots(4, 2, figsize=(18, 24))

# Chart 1: Customer Status (Pie)
axes[0, 0].pie(df_status['Count'], labels=df_status['Status'], autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
axes[0, 0].set_title('Figure 2.1: Customer Status Distribution')

# Chart 2: Customer Type (Bar)
sns.barplot(x='Type', y='Count', data=df_type, ax=axes[0, 1], palette='viridis')
axes[0, 1].set_title('Figure 2.2: Customers by Type')

# Chart 3: Gender Distribution (Pie)
axes[1, 0].pie(df_gender['Count'], labels=df_gender['Gender'], autopct='%1.1f%%', startangle=140, colors=sns.color_palette('husl'))
axes[1, 0].set_title('Figure 3.1: Gender Distribution')

# Chart 4: Age Brackets (Histogram/Bar)
sns.barplot(x='Age_Group', y='Count', data=df_age, ax=axes[1, 1], palette='coolwarm')
axes[1, 1].set_title('Figure 3.2: Customer Age Brackets')

# Chart 5: Geographic Distribution (Horizontal Bar)
sns.barplot(x='Count', y='Country', data=df_geo, ax=axes[2, 0], palette='flare')
axes[2, 0].set_title('Figure 4.1: Top 10 Countries')

# Chart 6: Income Band (Pie)
axes[2, 1].pie(df_income['Count'], labels=df_income['Band'], autopct='%1.1f%%', startangle=140, colors=sns.color_palette('Set2'))
axes[2, 1].set_title('Figure 5.2: Income Band Distribution')

# Chart 7: ID Document Types (Bar)
sns.barplot(x='Doc_Type', y='Count', data=df_id, ax=axes[3, 0], palette='magma')
axes[3, 0].set_title('Figure 6.2: Proof of Identity Types')
axes[3, 0].tick_params(axis='x', rotation=45)

# Chart 8: Profile Completeness (Bar/Histogram)
sns.barplot(x='Score', y='Count', data=df_complete, ax=axes[3, 1], palette='Blues_r')
axes[3, 1].set_title('Figure 9.1: Profile Completeness Scores')

plt.tight_layout()
plt.show()mport matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Configuration for visual style
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 10})

# --- DATA PREPARATION (Based on Report Pages 4-23) ---

# 1. Customer Status Distribution (Figure 2.1)
df_status = pd.DataFrame({
    'Status': ['ACTIVE', 'SUSPENDED', 'PENDING', 'DORMANT', 'CLOSED'],
    'Count': [98420, 18340, 12760, 8410, 4450]
})

# 2. Customer Type Distribution (Figure 2.2)
df_type = pd.DataFrame({
    'Type': ['INDIVIDUAL', 'CORPORATE', 'SME', 'NONPROFIT', 'GOVERNMENT'],
    'Count': [88500, 27400, 16900, 6400, 3180]
})

# 3. Gender Distribution (Figure 3.1)
df_gender = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Non-Binary', 'Prefer Not to Say'],
    'Count': [67840, 58920, 9240, 6380]
})

# 4. Age Group Distribution (Figure 3.2)
df_age = pd.DataFrame({
    'Age_Group': ['Under 18', '18-25', '26-35', '36-45', '46-60', '60+'],
    'Count': [2840, 24600, 42100, 31800, 27400, 13640]
})

# 5. Top 10 Countries by Count (Figure 4.1)
df_geo = pd.DataFrame({
    'Country': ['USA', 'UK', 'India', 'Germany', 'Canada', 'Australia', 'France', 'Brazil', 'Japan', 'South Africa'],
    'Count': [38200, 19400, 17600, 11300, 9800, 7400, 6900, 6100, 5200, 4100]
})

# 6. Income Band Distribution (Figure 5.2)
df_income = pd.DataFrame({
    'Band': ['< $25k', '$25k–$50k', '$50k–$100k', '$100k–$200k', '$200k+'],
    'Count': [18400, 31200, 42600, 21800, 10600]
})

# 7. Proof of Identity Types (Figure 6.2)
df_id = pd.DataFrame({
    'Doc_Type': ['Passport', 'Driver License', 'Utility Bill', 'Bank Statement', 'Gov. Letter'],
    'Count': [41200, 34600, 22400, 18900, 9280]
})

# 8. Profile Completeness Scores (Figure 9.1)
df_complete = pd.DataFrame({
    'Score': ['1 (Very Low)', '2', '3', '4', '5', '6 (Complete)'],
    'Count': [4200, 8400, 18600, 32400, 41800, 36980]
})

# --- PLOTTING ALL CHARTS ---
fig, axes = plt.subplots(4, 2, figsize=(18, 24))

# Chart 1: Customer Status (Pie)
axes[0, 0].pie(df_status['Count'], labels=df_status['Status'], autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
axes[0, 0].set_title('Figure 2.1: Customer Status Distribution')

# Chart 2: Customer Type (Bar)
sns.barplot(x='Type', y='Count', data=df_type, ax=axes[0, 1], palette='viridis')
axes[0, 1].set_title('Figure 2.2: Customers by Type')

# Chart 3: Gender Distribution (Pie)
axes[1, 0].pie(df_gender['Count'], labels=df_gender['Gender'], autopct='%1.1f%%', startangle=140, colors=sns.color_palette('husl'))
axes[1, 0].set_title('Figure 3.1: Gender Distribution')

# Chart 4: Age Brackets (Histogram/Bar)
sns.barplot(x='Age_Group', y='Count', data=df_age, ax=axes[1, 1], palette='coolwarm')
axes[1, 1].set_title('Figure 3.2: Customer Age Brackets')

# Chart 5: Geographic Distribution (Horizontal Bar)
sns.barplot(x='Count', y='Country', data=df_geo, ax=axes[2, 0], palette='flare')
axes[2, 0].set_title('Figure 4.1: Top 10 Countries')

# Chart 6: Income Band (Pie)
axes[2, 1].pie(df_income['Count'], labels=df_income['Band'], autopct='%1.1f%%', startangle=140, colors=sns.color_palette('Set2'))
axes[2, 1].set_title('Figure 5.2: Income Band Distribution')

# Chart 7: ID Document Types (Bar)
sns.barplot(x='Doc_Type', y='Count', data=df_id, ax=axes[3, 0], palette='magma')
axes[3, 0].set_title('Figure 6.2: Proof of Identity Types')
axes[3, 0].tick_params(axis='x', rotation=45)

# Chart 8: Profile Completeness (Bar/Histogram)
sns.barplot(x='Score', y='Count', data=df_complete, ax=axes[3, 1], palette='Blues_r')
axes[3, 1].set_title('Figure 9.1: Profile Completeness Scores')

plt.tight_layout()
plt.show()
```
